```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.neighbors import KNeighborsClassifier
```

```python
colors = pd.read_csv('color_samples.csv')
colors.head()
```

```python
sns.scatterplot(x='Brightness', labels='Saturation', hue='Class', data=colors, palette={'Red': 'red', 'Blue': 'blue'}, s=100)
plt.title('Brightness vs Saturation')
plt.grid(True)
plt.show()
```

```python
new_color = [40, 40]
distances = []
for brightness, saturation in zip(colors['Brightness'], colors['Saturation']):
    brightness_difference = brightness - new_color[0]
    saturation_difference = saturation - new_color[1]
    distance = np.sqrt(brightness_difference ** 2 + saturation_difference ** 2)
    distances.append(distance)
colors['Distance'] = distances
colors.sort_values('Distance')
```

```python
inputs = colors[['Brightness', 'Saturation']]
labels = colors['Class']
model = KNeighborsClassifier(n_neighbors=3, metric='euclidean')
model.fit(inputs, labels)
sample = pd.DataFrame([new_color], columns=inputs.columns)
result = model.predict(sample)[0]
print("Predicted class for", new_color, "with k=3:", result)
```

```python
plt.figure(figsize=(8, 6))
sns.scatterplot(x='Brightness', labels='Saturation', hue='Class', data=colors, palette={'Red': 'red', 'Blue': 'blue'}, s=100)
plt.scatter(new_color[0], new_color[1], color='green', marker='*', s=300, label='New Entry (40, 40)')
plt.title('KNN Classification')
plt.legend()
plt.grid(True)
plt.show()
```
