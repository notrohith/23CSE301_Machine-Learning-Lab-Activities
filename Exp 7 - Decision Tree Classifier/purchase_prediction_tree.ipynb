```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn import metrics, tree
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv('purchase_data.csv')

print("Dataset Shape:", data.shape)
display(data.head())
```

```python
columns = ['Age', 'EstimatedSalary']
inputs = data[columns]
labels = data['Purchased']

train_inputs, test_inputs, train_labels, test_labels = train_test_split(inputs, labels, test_size=0.2, random_state=42)

tree_model = DecisionTreeClassifier(criterion='entropy', max_depth=3, random_state=42)
tree_model.fit(train_inputs, train_labels)

predictions = tree_model.predict(test_inputs)
accuracy = metrics.accuracy_score(test_labels, predictions)
print("Accuracy Score:", accuracy)
```

```python
plt.figure(figsize=(16, 10))
tree.plot_tree(tree_model,
               feature_names=['Age', 'EstimatedSalary'],
               class_names=['Not Purchased', 'Purchased'],
               filled=True,
               rounded=True,
               fontsize=10)
plt.title("Decision Tree: Predicting Purchase based on Age and Salary")
plt.show()
```
