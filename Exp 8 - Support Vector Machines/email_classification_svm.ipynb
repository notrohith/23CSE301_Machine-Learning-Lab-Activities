```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score
import seaborn as sns
import matplotlib.pyplot as plt

%matplotlib inline
```

```python
emails = pd.read_csv('email_data.csv')
emails.head()
```

```python
emails.info()
print("\nMissing values:")
print(emails.isnull().sum())
print("\nEmail labels (y=spam, n=not spam):")
print(emails['yesno'].value_counts())
```

```python
labels = emails['yesno'].map({'y': 1, 'n': 0})
inputs = emails.drop(columns=['yesno'])

train_inputs, test_inputs, train_labels, test_labels = train_test_split(inputs, labels, test_size=0.3, random_state=42)

scaler = StandardScaler()
train_inputs = scaler.fit_transform(train_inputs)
test_inputs = scaler.transform(test_inputs)
```

```python
linear_svm = SVC(kernel='linear', random_state=0)
linear_svm.fit(train_inputs, train_labels)

linear_predictions = linear_svm.predict(test_inputs)

print("Accuracy (Linear Kernel):", accuracy_score(test_labels, linear_predictions))
print("\nClassification Report:\n", classification_report(test_labels, linear_predictions))
```

```python
linear_matrix = confusion_matrix(test_labels, linear_predictions)
sns.heatmap(linear_matrix, annot=True, fmt='d', cmap='Blues')
plt.title('SVM (Linear Kernel) Confusion Matrix')
plt.xlabel('Predicted Label')
plt.ylabel('Actual Label')
plt.show()
```

```python
rbf_svm = SVC(kernel='rbf', random_state=0)
rbf_svm.fit(train_inputs, train_labels)

rbf_predictions = rbf_svm.predict(test_inputs)

print("Accuracy (RBF Kernel):", accuracy_score(test_labels, rbf_predictions))
print("\nClassification Report:\n", classification_report(test_labels, rbf_predictions))
```

```python
rbf_matrix = confusion_matrix(test_labels, rbf_predictions)
sns.heatmap(rbf_matrix, annot=True, fmt='d', cmap='Oranges')
plt.title('SVM (RBF Kernel) Confusion Matrix')
plt.xlabel('Predicted Label')
plt.ylabel('Actual Label')
plt.show()
```
